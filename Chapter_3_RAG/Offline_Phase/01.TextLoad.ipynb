# 文档加载器

## 用于将不同格式的文件转化为字符串，方便后续切分以及向量化

## 1.PyPDFLoader

### 问题: pdf内部的表格超文本数据被破坏，无法正确加载

```python
from langchain_community.document_loaders import PyPDFLoader


# 初始化并配置文档加载器
loader = PyPDFLoader(
    file_path=r"C:\Users\55242\Desktop\组会\论文原文\On_Inter-Dataset_Code_Duplication_and_Data_Leakage_in_Large_Language_Models.pdf",  # 文件路径
    mode="page"  # page:按照PDF的自然页进行划分 single:整个PDF作为一页
)

# 加载PDF文件，得到list[document]
docs = loader.load()
print(f"文档数量{len(docs)}")

# 遍历，得到每一个document，都包含page_content和metadata两个属性
# enumerate() 函数用于将一个可遍历的数据对象(如列表、元组或字符串)组合为一个索引序列，同时列出数据和数据下标
for i, doc in enumerate(docs):
    print(f"文档编号:{i}")
    print(f"文档内容:{doc.page_content}")
    print(f"文档元数据:{doc.metadata}")
```

## 2.MinerU Loader

### 环境准备

```python
from mineru import MinerU
import os
from dotenv import load_dotenv
load_dotenv()
```

### 初始化客户端

```python
client = MinerU(token=os.getenv("MINERU_TOKEN"))
```

### 解析模式

```python
# 轻量解析
result = client.flash_extract("file_path")

print(result.markdown)
```

```python
# 精准解析
result = client.extract("file_path")

print(result.markdown)
```

```python
# OCR解析
result = client.extract("file_path", ocr=True)

print(result.markdown)
```
