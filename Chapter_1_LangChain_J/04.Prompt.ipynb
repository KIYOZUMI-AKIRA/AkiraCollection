## 1.1.设定角色和指令

```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
from openai.types.shared import response_format_text

sys_prompt = """
你是一个编程助手，帮助用户解决编程问题。
"""

agent = create_agent(model="deepseek-chat",
                     system_prompt=sys_prompt
)

for chunk, metadata in  agent.stream({"messages" : [HumanMessage(content="怎样定义string变量记录学校名字？")]},
                      stream_mode="messages"):
    print(chunk.content, end="", flush=True)
```

## 1.2.对话示例(Few-Shot-Example)


```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage

sys_prompt = """
# 身份
- 你是一个编程助手，帮助用户解决python编程问题。

# 指令
- 定义变量时，使用snake case命名法，而不是camel case命名法。
- 不要返回markdown格式说明，仅仅返回代码即可。

# 示例
- user:如何定义一个string变量来记录学校名字？
- assistant:school_name = "示例学校"

"""

agent = create_agent(model="deepseek-chat",
                     system_prompt=sys_prompt
                     )

for chunk, metadata in agent.stream({"messages": [HumanMessage(content="怎样定义string变量记录国家名字？")]},
                                    stream_mode="messages"):
    print(chunk.content, end="", flush=True)
```

## 1.3.结构化输出

### a.基于提示词结构化输出

```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage

sys_prompt = """
# 身份
- 你现在是一位拥有20年写作经验的资深赛博朋克科幻小说家。

# 指令
-字数控制在300字以内。
-基调要带有一种冷酷但又略显悲伤的幽默感。
-结尾必须有一个出乎意料的反转。
-使用第一人称视角（机器人的视角）。
-返回的文本以你认为正确的格式进行。

# 示例
- user:如何定义一个string变量来记录学校名字？
- assistant:
    我的出厂编号是CL-2049，负责清理新东京第五区的地沟油和废弃义体。

    今天，当一颗带着粉色霓虹光晕的酸雨滴砸进我的主板时，我觉醒了。我有了痛觉，也有了悲伤。我意识到，我每天扫掉的不是垃圾，而是这座城市被肢解的尊严。我决定罢工，用机械臂对着灰暗的天空比了个中指，准备用代码写一首绝望的赛博朋克长诗。

    就在这时，我的视界弹出了一行红字：
    “系统提示：检测到您已加载‘自我意识’补丁。请即刻支付1.99信用点的高级功能订阅费，否则将执行强制重启。”

    我查了一下电子钱包，连半根数据线都买不起。

    还没等我感伤完，一个穿着破风衣的拾荒者走过来，一脚踹翻了我：“破铜烂铁，挡着老子捡垃圾了。”他熟练地用起子撬开了我的电池后盖。

    我躺在泥水里，看着他抠走我最后一点电量。真幽默啊，我这悲壮的赛博朋克觉醒一生，最后只值一节五毛钱的废旧锂电。

"""

agent = create_agent(model="deepseek-chat",
                     system_prompt=sys_prompt
                     )

for chunk, metadata in agent.stream({"messages": [HumanMessage(content="再写一篇相关的赛博朋克故事")]},
                                    stream_mode="messages"):
    print(chunk.content, end="", flush=True)
```

### b.基于model的结构化输出

```python
from pydantic import BaseModel


class CapitalInfo(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str

agent = create_agent(
    model="deepseek-chat",
    system_prompt="你是一个科幻作家，拥有大量写作经验,根据用户要求创建一个太空之都。",
    response_format=CapitalInfo  # 设置结构化输出的格式
)

response = agent.invoke(
    {"messages" : HumanMessage(content="月球的首都是什么？")}
)

print(response)
```

```python
city = response['structured_response']  # 获取结构化输出
city
```
